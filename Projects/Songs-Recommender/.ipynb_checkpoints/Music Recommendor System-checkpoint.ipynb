```python
import numpy as np
import pandas as pd
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("spotify_2023.csv", encoding="ISO-8859-1")
```

```python
data.info()
```

```python
features = ["bpm", "danceability_%", "energy_%", "acousticness_%", "instrumentalness_%"]
```

```python
X = data[features]
```

```python
X = X.fillna(X.mean())
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
knn = NearestNeighbors(n_neighbors=10, metric="euclidean")
knn.fit(X_scaled)
```

```python
random_songs = data.sample(n=10, random_state=537)
```

```python
for _, song in random_songs.iterrows():
    song_name = song["track_name"]
    print(f"\nSelected Song: {song_name}")

    song_index = data[data["track_name"] == song_name].index[0]
    query_song = X.iloc[song_index].values.reshape(1, -1)

    distances, indices = knn.kneighbors(query_song)
    recommended_songs = data.iloc[indices[0]]

    print(f"Similar songs to {song_name}:")
    print(recommended_songs[["track_name", "artist(s)_name"]])
```
