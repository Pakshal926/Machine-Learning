```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("uber_stock_data.csv")
df.head()
```

```python
df.info()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.info()
```

```python
plt.style.use('fivethirtyeight')
plt.figure(figsize=(16,6))
plt.plot(df["Date"] , df["Close"])
plt.title("UBER STOCK PRICE CHART")
plt.xlabel("Date")
plt.ylabel("Price(USD)")
plt.show()
```

```python
yillik=df.groupby(pd.Grouper(key="Date" , freq="Y")).mean()
yillik.index=yillik.index.year
```

```python
yillik
```

```python
plt.bar(yillik.index, yillik["Close"], color = ["red", "blue"])
plt.title("Stock value per year")
plt.show()
```
